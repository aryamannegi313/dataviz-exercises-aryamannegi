```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('/content/co2_emissions.csv')

print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")

print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country/countries **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country/countries and its story

```python

# Task 1: CO2 emissions per capita

highlight = 'China'

fig = go.Figure()

# Draw all other countries in grey
for country in df['Country'].unique():

    if country != highlight:

        country_data = df[
            df['Country'] == country
        ].sort_values('Year')

        fig.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_per_capita'],
            mode='lines',
            name=country,
            line=dict(color='#EEEEEE', width=1.5)
        ))

# Select China
china = df[
    df['Country'] == highlight
].sort_values('Year')

# Draw China in blue
fig.add_trace(go.Scatter(
    x=china['Year'],
    y=china['CO2_per_capita'],
    mode='lines',
    name='China',
    line=dict(color='#2563EB', width=4)
))

# Label China directly at the end
fig.add_annotation(
    x=china['Year'].iloc[-1],
    y=china['CO2_per_capita'].iloc[-1],
    text='China',
    showarrow=False,
    xanchor='left',
    xshift=8,
    font=dict(color='#2563EB', size=15)
)

# Calculate the change over time
start = china['CO2_per_capita'].iloc[0]
end = china['CO2_per_capita'].iloc[-1]

direction = (
    'increased' if end > start
    else 'decreased' if end < start
    else 'remained unchanged'
)

# Improve the chart design
fig.update_layout(
    title=f"China's CO2 emissions per capita {direction} from 2000 to 2022",
    xaxis_title='Year',
    yaxis_title='CO2 per Capita',
    template='plotly_white',
    showlegend=False,
    font=dict(family='Arial', size=14),
    xaxis=dict(range=[2000, 2025]),
    height=550
)

fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python

# Task 2: CO2 emissions in Asian countries

# Select only Asian countries
asia = df[df['Region'] == 'Asia']

highlight = 'India'

fig2 = go.Figure()

# Draw other Asian countries in grey
for country in asia['Country'].unique():

    if country != highlight:

        country_data = asia[
            asia['Country'] == country
        ].sort_values('Year')

        fig2.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_Mt'],
            mode='lines',
            name=country,
            line=dict(color='#EEEEEE', width=1.5)
        ))

# Select India
india = asia[
    asia['Country'] == highlight
].sort_values('Year')

# Highlight India in orange
fig2.add_trace(go.Scatter(
    x=india['Year'],
    y=india['CO2_Mt'],
    mode='lines',
    name='India',
    line=dict(color='#D97706', width=4)
))

# Add direct label at the end
fig2.add_annotation(
    x=india['Year'].iloc[-1],
    y=india['CO2_Mt'].iloc[-1],
    text='India',
    showarrow=False,
    xanchor='left',
    xshift=8,
    font=dict(color='#D97706', size=15)
)

# Calculate percentage change
start = india['CO2_Mt'].iloc[0]
end = india['CO2_Mt'].iloc[-1]

change = ((end - start) / start) * 100

# Improve chart design
fig2.update_layout(
    title=f"India's CO2 emissions changed by {change:+.1f}% from 2000 to 2022",
    xaxis_title='Year',
    yaxis_title='CO2 Emissions (Million Tonnes)',
    template='plotly_white',
    showlegend=False,
    font=dict(family='Arial', size=14),
    xaxis=dict(range=[2000, 2025]),
    height=550
)

fig2.show()
```
